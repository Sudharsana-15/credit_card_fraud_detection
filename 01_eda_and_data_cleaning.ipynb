```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

print('Libraries loaded successfully!')
```

### 1. Data Ingestion & Audit (Fry's Stage 1 & 2: Acquire & Parse)


```python
raw_csv = r'../data/raw/creditcard.csv'
df = pd.read_csv(raw_csv)
print('Raw Shape:', df.shape)
print('Total Missing Values:', df.isnull().sum().sum())
print('Duplicate Rows:', df.duplicated().sum())
```

### 2. Duplicate Inspection & Filtering (Fry's Stage 3: Filter)
Inspecting duplicate transactions before removal to prevent data leakage between train and test sets.


```python
dup_redundant = df.duplicated(keep='first')
print('Redundant duplicates by class:')
print(df[dup_redundant]['Class'].value_counts())

# Drop redundant duplicates
df_clean = df.drop_duplicates(keep='first').copy()
print('Cleaned Shape:', df_clean.shape)
```

### 3. Statistical Analysis & Mining (Fry's Stage 4: Mine)
Discrete Bernoulli model of fraud, summary statistics, and correlations.


```python
p_fraud = (df_clean['Class'] == 1).mean()
print(f'Bernoulli Parameter p (Fraud Rate): {p_fraud:.6f} ({p_fraud*100:.4f}%)')
print(f'Theoretical Variance Var(Y) = p*(1-p): {p_fraud*(1-p_fraud):.6f}')

stats_cols = ['Amount', 'Time', 'V1', 'V2', 'V14', 'V17']
print(df_clean[stats_cols].describe().T[['mean', 'std', 'min', '50%', 'max']])
```
